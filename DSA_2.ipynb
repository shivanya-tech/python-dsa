# Searching and Sorting Algorithms
This notebook contains implementations of Linear Search, Binary Search (recursive and iterative), Selection Sort, Insertion Sort, and Merge Sort.

```python
# Linear Search
def linear_serach(l,v):
    for i in l:
        if i==v:
            return "found"
    return "not found"
l=[1,2,3,5,7]
v=3
print(linear_serach(l,v))
```

```python
# Binary Search (recursive)
def binarysearch(n,l):
    if l == []:
        return False
    length=len(l)
    mid=length//2
    if n==l[mid]:
        return True
    if n< l[mid]:
        return binarysearch(n,l[0:mid])
    else:
        return binarysearch(n,l[mid+1:])

print(binarysearch(7,[0,1,2,4,5,9]))
```

```python
# Optimized Binary Search (recursive with indices)
def b_sea(l, v, left=0, right=None):
    if right is None:
        right = len(l) - 1
    if left > right:
        return False
    mid = (left + right) // 2
    if l[mid] == v:
        return True
    elif v < l[mid]:
        return b_sea(l, v, left, mid - 1)
    else:
        return b_sea(l, v, mid + 1, right)

l = [1, 2, 3, 5, 7]
v = 9
print(b_sea(l, v))  # Output: False
```

```python
# Iterative Binary Search
def binary_search(l, v):
    n = len(l)
    left = 0
    right = n - 1
    while left <= right:
        mid = (left + right) // 2
        if v == l[mid]:
            return True
        elif v < l[mid]:
            right = mid - 1
        else:
            left = mid + 1
    return False

l = [1, 2, 3, 4, 5]
print(binary_search(l, 6))  # Output: False
```

```python
# Selection Sort (basic)
def selection_sort(l):
    n=len(l)
    for i in range(n):
        for j in range(i+1,n):
            if l[i]>l[j]:
                l[i],l[j]=l[j],l[i]
    return l

l=[1,4,9,2,3,11]
print(selection_sort(l))
```

```python
# Selection Sort (efficient)
def selection_sort(l):
    n = len(l)
    if n < 1:
        return l
    if n > 1:
        for i in range(n):
            mpos = i
            for j in range(i + 1, n):
                if l[j] < l[mpos]:
                    mpos = j
            l[i], l[mpos] = l[mpos], l[i]
    return l

print(selection_sort([2,5,1,4,9,0]))
```

```python
# Insertion Sort
def insertion_sort(l):
    n=len(l)
    for i in range(n):
        j=i
        while (j>0 and l[j]<l[j-1]):
            l[j],l[j-1]=l[j-1],l[j]
            j-=1
    return l

l=[2,5,1,4,9,0]
print(insertion_sort(l))
```

```python
# Merge Sort
def merge_sort(arr):
    n=len(arr)
    if n<=1:
        return arr
    m=n//2
    left=merge_sort(arr[0:m])
    right=merge_sort(arr[m:])
    return merge(left, right)


def merge(left,right):
    result=[]
    i=j=0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

arr = [38, 27, 43, 3, 9, 82, 10]
sorted_arr = merge_sort(arr)
print("Sorted array:", sorted_arr)
```
