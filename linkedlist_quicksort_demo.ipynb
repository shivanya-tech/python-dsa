```python
# Quicksort Implementation
def quicksort(L, f, r):
    if r - f <= 1:
        return
    pivot = L[f]
    lower = f + 1
    for i in range(f + 1, r):
        if L[i] <= pivot:
            L[i], L[lower] = L[lower], L[i]
            lower += 1
    L[f], L[lower - 1] = L[lower - 1], L[f]
    quicksort(L, f, lower - 1)
    quicksort(L, lower, r)
    return L

# Example
L = [7, 2, 9, 4, 3]
print("Quicksort:", quicksort(L, 0, len(L)))
```

```python
# Node class
class Node:
    def __init__(self, data=None):
        self.data = data
        self.next = None

# LinkedList class
class LinkedList:
    def __init__(self):
        self.head = None

    def isempty(self):
        return self.head is None

    def append(self, value):
        newnode = Node(value)
        if self.isempty():
            self.head = newnode
            return
        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = newnode

    def insert(self, value):
        newnode = Node(value)
        newnode.next = self.head
        self.head = newnode

    def delete(self, value):
        if self.isempty():
            return
        if self.head.data == value:
            self.head = self.head.next
            return
        temp = self.head
        while temp.next and temp.next.data != value:
            temp = temp.next
        if temp.next:
            temp.next = temp.next.next

    def reverse(self):
        prev, current = None, self.head
        while current:
            nxt = current.next
            current.next = prev
            prev = current
            current = nxt
        self.head = prev

    def print_list(self):
        current = self.head
        while current:
            print(current.data, end=" --> ")
            current = current.next
        print("None")
```

```python
# Example Usage
ll = LinkedList()
ll.insert(3)
ll.insert(2)
ll.insert(1)
print("After inserts:")
ll.print_list()

ll.append(4)
ll.append(5)
print("After appends:")
ll.print_list()

ll.delete(3)
print("After deleting 3:")
ll.print_list()

ll.reverse()
print("After reversing:")
ll.print_list()
```

```python
# Stack Implementation using Python List
class Stack:
    def __init__(self):
        self.items = []

    # Check if stack is empty
    def is_empty(self):
        return len(self.items) == 0

    # Push element onto stack
    def push(self, item):
        self.items.append(item)

    # Pop element from stack
    def pop(self):
        if self.is_empty():
            print("Stack is empty!")
            return None
        return self.items.pop()

    # Peek top element
    def peek(self):
        if self.is_empty():
            print("Stack is empty!")
            return None
        return self.items[-1]

    # Get size
    def size(self):
        return len(self.items)

    # Print stack
    def print_stack(self):
        print(self.items)
```

```python
s = Stack()
s.push(10)
s.push(20)
s.push(30)
print("Stack after pushes:")
s.print_stack()

print("Top element:", s.peek())
print("Popped:", s.pop())
print("Stack after pop:")
s.print_stack()
```

```python
# Queue Implementation using Python List
class Queue:
    def __init__(self):
        self.items = []

    # Check if queue is empty
    def is_empty(self):
        return len(self.items) == 0

    # Enqueue (add element at rear)
    def enqueue(self, item):
        self.items.append(item)

    # Dequeue (remove element from front)
    def dequeue(self):
        if self.is_empty():
            print("Queue is empty!")
            return None
        return self.items.pop(0)

    # Peek front element
    def peek(self):
        if self.is_empty():
            print("Queue is empty!")
            return None
        return self.items[0]

    # Get size
    def size(self):
        return len(self.items)

    # Print queue
    def print_queue(self):
        print(self.items)
```

```python
q = Queue()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
print("Queue after enqueues:")
q.print_queue()

print("Front element:", q.peek())
print("Dequeued:", q.dequeue())
print("Queue after dequeue:")
q.print_queue()
```
